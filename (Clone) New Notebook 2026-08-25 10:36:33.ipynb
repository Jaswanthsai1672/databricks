```python
from pyspark.sql.functions import col,lit,count,avg,max,min,sum
from pyspark.sql import SparkSession

# spark = SparkSession.builder \
#        .appName("Practice") \
#        .getOrCreate

data = [(1, "Jaswanth","IT", 200000),
        (2, "Rahul", "HR", 50000),
        (3, "Rohan", "IT", 100000)]
df = spark.createDataFrame(
    data, ["id", "name","DEPARTMENT", "salary"]
)
df.show()
```

```python
rahul_df = df.filter(col("name") == "Rahul")

rahul_df = rahul_df.withColumn("salary", lit(70000)).show()
```

```python
from pyspark.sql.functions import when

df.withColumn(
    "salary_category",
    when(col("salary") >= 50000, "Low")
    .otherwise("High")
).show()
```

```python
df.groupBy("department").agg(
    count("*").alias("employee_count"),
    avg("salary").alias("avg_salary"),
    max("salary").alias("max_salary"),
    min("salary").alias("min_salary")
).show()
```
