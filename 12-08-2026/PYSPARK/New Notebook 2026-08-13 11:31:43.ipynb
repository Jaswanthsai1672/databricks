```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql.window import Window
spark = SparkSession.builder.getOrCreate()

data = [
    (1, "Ravi",   "IT",      60000, "2024-01-10"),
    (2, "Arun",   "IT",      80000, "2024-02-15"),
    (3, "Kiran",  "IT",      70000, "2024-03-20"),
    (4, "Suresh", "IT",      80000, "2024-04-10"),
    (5, "Priya",  "HR",      50000, "2024-01-12"),
    (6, "Sita",   "HR",      60000, "2024-02-18"),
    (7, "Anu",    "HR",      60000, "2024-03-25"),
    (8, "Rahul",  "Finance", 90000, "2024-01-05"),
    (9, "Vijay",  "Finance", 70000, "2024-02-10"),
    (10,"Mohan",  "Finance", 70000, "2024-03-15")
]

df = spark.createDataFrame(
    data,
    ["id", "name", "department", "salary", "joining_date"]
)

df = df.withColumn(
    "joining_date",
    to_date("joining_date")
) 
df.show()
```

```python
window_spec = Window.orderBy(
    col("salary").desc()
)
result = df.withColumn(
    "row_num",
    row_number().over(window_spec)
)
result.show()
```

```python
window_spec = Window.orderBy(
    col("salary").desc()
)
result = df.withColumn(
    "salary_rank",
    rank().over(window_spec)
)
result.show()
```

```python
window_spec = Window.orderBy(
    col("salary").desc()
)
result = df.withColumn(
    "salary_rank",
    dense_rank().over(window_spec)

)
result.show()
```

```python
window_spec = Window.orderBy(
    col("salary").desc()

)
result = (
    df.withColumn(
        "row_number",
        row_number().over(window_spec)

    )
    .withColumn(
        "rank",
        rank().over(window_spec)

    )
    .withColumn(
        "dense_rank",
        dense_rank().over(window_spec)

    )

)
result.orderBy(col("salary").desc()).show()
```

```python
window_spec = Window.partitionBy(
    "department"

).orderBy(
    col("salary").desc()

)
result = df.withColumn(
    "dept_rank",
    dense_rank().over(window_spec)
)
result.show()
```

```python
window_spec = Window.partitionBy(
    "department"
).orderBy(
    col("salary").desc()
)

ranked_df = df.withColumn(
    "rank",
    dense_rank().over(window_spec)
)

result = ranked_df.filter(
    col("rank") <= 2
)

result.show()
```

```python
window_spec = Window.partitionBy(
    "department"
).orderBy(
    col("salary")

)
result = df.withColumn(
    "previous_salary",
    lag("salary").over(window_spec)
)
result.show()
```

```python
window_spec = Window.partitionBy(
    "department"
).orderBy(
    col("salary")
)

result = (
    df
    .withColumn(
        "previous_salary",
        lag("salary").over(window_spec)
    )
    .withColumn(
        "salary_difference",
        col("salary") - col("previous_salary")
    )
)

result.show()
```

```python
window_spec = Window.orderBy(
    col("salary")
)
result = df.withColumn(
    "next_salary",
    lead("salary").over(window_spec)
)
result.show()
```

```python
window_spec = Window.orderBy(
    col("salary")
)
result = (df.withColumn(
    "previous_salary",
    lag("salary").over(window_spec)
)
.withColumn(
    "next_salary",
    lead("salary").over(window_spec)
)
)
result.show()
```

```python
window_spec = Window.orderBy(
    "joining_date"
).rowsBetween(
    Window.unboundedPreceding,
    Window.currentRow
)
result = df.withColumn(
    "running_salary",
    sum("salary").over(window_spec)
)
result.show()
```
