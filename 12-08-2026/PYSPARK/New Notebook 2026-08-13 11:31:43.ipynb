```python

# 1.Create a DataFrame from a list of tuples. Print schema, display data, rename Salary to MonthlySalary.
# SparkSession is already available as 'spark' in Databricks notebooks

data = [
    (101, "Jaswanth", 500000.0),
    (102, "Praneeth", 6000.0),
    (103, "Charan", 55000.0),
    (104, "Sai", 70000.0)
]
df = spark.createDataFrame(
    data,
    ["EmployeeId", "EmployeeName", "Salary"]
)

df.printSchema()
df.show()
df = df.withColumnRenamed(
    "Salary",
    "MonthlySalary"
)
df.show()

from pyspark.sql import SparkSession
from pyspark.sql.functions import lit

spark = SparkSession.builder \
    .appName("Spark DataFrames") \
    .getOrCreate()

data = [
    {"EmployeeID": 101, "Name": "Jaswanth", "Salary": 100000},
    {"EmployeeID": 102, "Name": "Praneeth", "Salary": 50000},
    {"EmployeeID": 103, "Name": "Charan","Salary": 70000},
    {"EmployeeID": 104, "Name": "Sai", "Salary": 80000}
]

df = spark.createDataFrame(data)
df.printSchema()
df.show()
df = df.withColumn(
    "Department",
    lit("IT")
)
df.show()
from pyspark.sql import SparkSession
from pyspark.sql.types import (
    StructType,
    StructField,
    IntegerType,
    StringType,
    DoubleType,
    DateType
)
from datetime import date

spark = SparkSession.builder \
    .appName("Spark DataFrames") \
    .getOrCreate()

data = [
    (101, "Jaswanth", 100000, date(2026, 8, 28)),
    (102, "Praneeth", 60000, date(2026, 9, 1)),
    (103, "Charan", 70000, date(2026, 9, 5)),
    (104, "Rohan", 80000, date(2026, 10, 2)),
    (105, "Sai", 90000, date(2026, 9, 5))
]

schema = StructType([
    StructField("EmployeeID", IntegerType(), True),
    StructField("Name", StringType(), True),
    StructField("Salary", DoubleType(), True),
    StructField("JoiningDate", DateType(), True)
])

df = spark.createDataFrame(data, schema)
df.printSchema()
df.show()

data = [(1,"Sai",25, "IT",80000),
        (2,"Ravi",26,"HR",70000),
        (3,"Rajesh",27,"IT",60000),
        (4,"Ramesh",28,"IT",50000),
        (5,"Rakesh",29,"HR",40000)]

columns = ["id","name","age","department","salary"]
df = spark.createDataFrame(data, columns)
display(df)
df.show()
display(df)
df.printSchema()
df.select("name", "salary").show()

display(df.select("name", "salary"))
df.filter(df.salary > 60000).show()
df.filter("salary > 60000").count()
df.filter("salary > 60000").collect()
df.filter("salary > 60000").take(1)
df.filter("salary > 60000").head()
df.filter("salary > 60000").first()
df.filter(df.age > 25).show()
df.filter(
    (df.department == "IT") &
    (df.age > 25)).show()
df.filter(
    (df.department == "IT") |
    (df.department == "HR")
).show()
from pyspark.sql.functions import col
df2 = df.withColumn(
    "annual_salary",
    col("salary") * 12
)
display(df2)
df2 = df.withColumnRenamed(
    "salary",
    "annual_salary"
)
display(df2)
df2 = df.drop("age")
display(df2)
df.orderBy("salary").show()
df.orderBy(col("salary").desc()).show()
df.count()
```
